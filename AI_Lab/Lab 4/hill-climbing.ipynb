<h2 class="section-title">Part 1: Background</h2>
<div class="card">
    <h3><span class="badge">1</span> What is Hill Climbing Search?</h3>
    <div class="theory-box">
        <p><strong>Definition:</strong> Hill Climbing is a local search algorithm that continuously moves in the direction of <strong>increasing value (uphill)</strong> to find the peak (maximum). It is called "greedy" because it always chooses the best immediate option without looking ahead.</p>
    </div>
    <div class="theory-box">
        <p><strong>Simple Analogy:</strong></p>
        <p>Imagine you are <strong>blindfolded on a mountain</strong> and want to reach the highest point. You can only feel the ground immediately around you. You take a step in the steepest upward direction. You repeat until you cannot go higher. This is exactly how hill climbing works!</p>
    </div>
    <div class="theory-box">
        <p><strong>Key Terminology:</strong></p>
        <ul>
            <li><strong>Current State:</strong> Where the agent is now</li>
            <li><strong>Objective Function:</strong> What we want to maximize (height, profit, score)</li>
            <li><strong>Neighbor State:</strong> A state reachable by one action</li>
            <li><strong>Local Maximum:</strong> A peak higher than neighbors but not the highest overall</li>
            <li><strong>Global Maximum:</strong> The highest peak overall (our true goal)</li>
        </ul>
    </div>
    <div class="tree"><strong>📊 State Space Diagram:</strong>
        <pre>Objective
  Value
    &uarr;
    |        /\    Global Maximum
    |       /  \      /\
    |      /    \    /  \
    |     /      \  /    \
    |    /        \/      \
    |   /        Local     \
    |  /        Maximum     \
    | /                      \
    |/                        \
    +--------------------------------&rarr; State Space

    Legend:
    * Local Maximum: A peak that is not the highest
    * Global Maximum: The highest peak overall
    * Current State: Where the agent is now</pre>
    </div>
</div>

<h2 class="section-title">Part 2: Algorithm and Pseudocode</h2>
<div class="card">
    <h3><span class="badge">2</span> Hill Climbing Algorithm</h3>
    <div class="step-box">
        <p><strong>Algorithm Steps (for Maximization):</strong></p>
        <ol>
            <li>Start with a <strong>current state</strong> (initial solution).</li>
            <li>Evaluate the objective function at the current state.</li>
            <li>Generate all <strong>neighbors</strong> of the current state.</li>
            <li>If a neighbor has a <strong>higher value</strong> than the current state:
                <ul>
                    <li>Move to that neighbor (make it the new current state).</li>
                    <li>Repeat from step 2.</li>
                </ul>
            </li>
            <li>If <strong>no neighbor is better</strong> &rarr; stop (reached a peak).</li>
            <li>Return the current state as the solution.</li>
        </ol>
    </div>
    <div class="step-box">
        <p><strong>Pseudocode:</strong></p>
        <div class="code-box">
            <pre>function HILL_CLIMBING(values, start_index):
    current = start_index
    
    while True:
        // Check left neighbor
        left_higher = False
        if current &gt; 0 and values[current - 1] &gt; values[current]:
            left_higher = True
        
        // Check right neighbor
        right_higher = False
        if current &lt; length(values) - 1 and values[current + 1] &gt; values[current]:
            right_higher = True
        
        // If no neighbor is higher, stop (peak reached)
        if not left_higher and not right_higher:
            return current
        
        // Move to the higher neighbor
        if left_higher and right_higher:
            if values[current - 1] &gt;= values[current + 1]:
                current = current - 1
            else:
                current = current + 1
        elif left_higher:
            current = current - 1
        else:
            current = current + 1</pre>
        </div>
    </div>
    <div class="step-box">
        <p><strong>Simple Explanation:</strong></p>
        <ul>
            <li>Start at a random point on the landscape.</li>
            <li>Look at the left and right neighbors.</li>
            <li>If a neighbor is higher, move to it.</li>
            <li>If both neighbors are higher, move to the highest one.</li>
            <li>If no neighbor is higher, stop &ndash; you have reached a peak.</li>
        </ul>
    </div>
</div>

    else:
        print("⚠ LOCAL MAXIMUM (not the highest)\n")
    
    return current
    # Step 4: Test from multiple starting positions

for start in start_positions:

<h2 class="section-title">Part 5: Problems with Hill Climbing</h2>
<div class="card">
    <h3><span class="badge">5</span> Limitations of Hill Climbing</h3>
    <div class="warning-box">
        <p>⚠️ Hill climbing has three major problems:</p>
    </div>
    <div class="theory-box">
        <p><strong>1. Local Maximum:</strong></p>
        <ul>
            <li><strong>What it is:</strong> A peak that is higher than its neighbors but NOT the highest peak overall.</li>
            <li><strong>Why it happens:</strong> The algorithm stops when no neighbor is better, even if a higher peak exists elsewhere.</li>
            <li><strong>Analogy:</strong> You climb a small mound and think you are at the top, but Mount Everest is behind you.</li>
            <li><strong>Solution:</strong> <strong>Random restart</strong> &ndash; run hill climbing multiple times from different starting points.</li>
        </ul>
    </div>
    <div class="theory-box">
        <p><strong>2. Plateau (Flat Area):</strong></p>
        <ul>
            <li><strong>What it is:</strong> A flat area where all neighbors have the same value.</li>
            <li><strong>Why it happens:</strong> The algorithm cannot decide which direction to go.</li>
            <li><strong>Analogy:</strong> You are on a vast flat field &ndash; every direction looks exactly the same.</li>
            <li><strong>Solution:</strong> <strong>Sideways moves</strong> &ndash; allow moving to states with equal value (limit the number to avoid infinite loops).</li>
        </ul>
    </div>
    <div class="theory-box">
        <p><strong>3. Ridge:</strong></p>
        <ul>
            <li><strong>What it is:</strong> A narrow peak where every small step seems to go down, but moving along the ridge goes up.</li>
            <li><strong>Why it happens:</strong> The algorithm only looks at immediate neighbors, which might be lower.</li>
            <li><strong>Analogy:</strong> Walking on a narrow wall &ndash; stepping left or right makes you fall, but moving forward keeps you on the ridge.</li>
            <li><strong>Solution:</strong> Move in <strong>multiple directions at once</strong> or use <strong>simulated annealing</strong>.</li>
        </ul>
    </div>
    <div class="tree"><strong>Visual Representation:</strong>
