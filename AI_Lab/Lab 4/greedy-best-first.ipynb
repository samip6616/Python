<h2 class="section-title">Part 1: Background</h2>
<div class="card">
    <h3><span class="badge">1</span> What is Greedy Best-First Search?</h3>
    <div class="theory-box">
        <p><strong>Definition:</strong> Greedy Best-First Search is an informed search algorithm that expands the node that appears to be <strong>closest to the goal</strong>, based solely on a heuristic function h(n).</p>
        <p><strong>Evaluation Function:</strong> f(n) = h(n)</p>
        <ul>
            <li><strong>h(n):</strong> Heuristic estimate of cost from node n to the goal</li>
            <li>It ignores the actual cost from the start (g(n))</li>
        </ul>
    </div>
    <div class="theory-box">
        <p><strong>Simple Analogy:</strong></p>
        <p>Imagine you are in Kathmandu and want to reach Pokhara. You ask a bird "which city looks closest to Pokhara?" The bird points to a city that looks close on the map, but the actual road might be long. Greedy BFS works the same way &ndash; it only looks at how close something <strong>seems</strong>, not the actual distance traveled.</p>
    </div>
    <div class="theory-box">
        <p><strong>Key Characteristics:</strong></p>
        <ul>
            <li>Uses a <strong>Priority Queue</strong> (ordered by h(n) value &ndash; smallest first)</li>
            <li><strong>Not complete</strong> &ndash; can get stuck in infinite loops</li>
            <li><strong>Not optimal</strong> &ndash; may not find the shortest path</li>
            <li><strong>Fast</strong> but can be misled by bad heuristics</li>
        </ul>
    </div>
    <div class="tree"><strong>🌲 Graph with Heuristic Values (h):</strong>
        <pre>        S (h=10)

    F (h=2) G (h=0)</pre>

<h2 class="section-title">Part 2: Algorithm</h2>
<div class="card">
    <h3><span class="badge">2</span> Greedy Best-First Search Algorithm</h3>
    <div class="step-box">
        <p><strong>Algorithm Steps:</strong></p>
        <ol>
            <li>Start with the initial node. Add it to a priority queue ordered by h(n).</li>
            <li>Create an empty <strong>visited list</strong> to keep track of nodes already expanded.</li>
            <li>While the priority queue is not empty:
                <ol type="a">
                    <li>Remove the node with the <strong>smallest h(n)</strong> value from the queue.</li>
                    <li>If this node is the <strong>goal</strong>, return the path.</li>
                    <li>If this node has already been visited, <strong>skip</strong> it.</li>
                    <li>Mark the current node as <strong>visited</strong>.</li>
                    <li>Generate all <strong>neighbors</strong> of the current node.</li>
                    <li>For each neighbor that has <strong>not been visited</strong>, add it to the queue with its h(n) value.</li>
                </ol>
            </li>
            <li>If the queue becomes empty, <strong>no solution exists</strong>.</li>
        </ol>
    </div>
    <div class="step-box">
        <p><strong>Pseudocode:</strong></p>
        <div class="code-box">
            <pre>function GREEDY_BFS(graph, start, goal, heuristic):
    queue = [(heuristic(start), start, [start])]
    visited = []

    while queue is not empty:
        sort queue by h(n)
        h_val, current, path = queue.pop(0)

        if current in visited:
            continue

        if current == goal:
            return path

        visited.append(current)

        for neighbor in graph[current]:
            if neighbor not in visited:
                new_path = path + [neighbor]
                add (heuristic(neighbor), neighbor, new_path) to queue

    return None  // No path found</pre>
        </div>
    </div>
</div>

<h2 class="section-title">Part 3: Code</h2>
<div class="card">
    <h3><span class="badge">3</span> Complete Python Code</h3>
    <div class="code-box">
        <pre># Lab: Greedy Best-First Search
# Program by: [Your Name]
# Roll no: [Your Roll No]
# Step 1: Define the Graph (connections between nodes)
graph = {
    'S': ['A', 'B'],
    'A': ['C', 'D'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['F', 'G'],
    'E': ['G'],
    'F': [],
    'G': []
}
# Step 2: Define Heuristic Values (estimated distance to goal G)
heuristic = {
    'S': 10,
    'A': 8,
    'B': 6,
    'C': 5,
    'D': 4,
    'E': 3,
    'F': 2,
    'G': 0
}
# Step 3: Greedy BFS Algorithm
def greedy_bfs(start, goal):
    # Queue stores (h_value, node, path)
    queue = [(heuristic[start], start, [start])]
    visited = []  # Keep track of visited nodes

    print("=" * 60)
    print("GREEDY BEST-FIRST SEARCH")
    print("=" * 60)
    print()
    print("Step | Current Node | h(n) | Path")
    print("-----|--------------|------|-----------------")

    step = 1

    while queue:
        # Sort by heuristic value (smallest first)
        queue.sort()
        
        # Get the node with smallest h(n)
        h_val, current, path = queue.pop(0)

        # Skip if already visited
        if current in visited:
            continue

        print(f"{step:4} | {current:12} | {h_val:4} | {' -&gt; '.join(path)}")
        step = step + 1

        # If goal found, return path
        if current == goal:
            print()
            print("=" * 60)
            print("GOAL FOUND!")
            print("=" * 60)
            print("Path:", ' -&gt; '.join(path))
            return path

        # Mark current as visited
        visited.append(current)

        # Add all unvisited neighbors to queue
        for neighbor in graph[current]:
            if neighbor not in visited:
                new_path = path + [neighbor]
                queue.append((heuristic[neighbor], neighbor, new_path))

    print("Goal not found!")
    return None

Step 4: Run the algorithm
greedy_bfs('S', 'G')
</pre>

<div class="card">
    <h3><span class="badge">A1</span> Question 1 Sample Answer</h3>
    <div class="step-box">
        <p>The evaluation function of Greedy Best-First Search is f(n) = h(n).</p>
        <p><strong>h(n):</strong> Heuristic estimate of the cost from node n to the goal (estimated distance remaining).</p>
    </div>
    <h3><span class="badge">A2</span> Question 2 Sample Answer</h3>
    <div class="step-box">
        <p>Greedy BFS found the path: <strong>S &rarr; B &rarr; D &rarr; G</strong></p>
    </div>
    <h3><span class="badge">A3</span> Question 3 Sample Answer</h3>
    <div class="step-box">
        <p>No, Greedy BFS is <strong>not guaranteed</strong> to find the shortest path. It only looks at the heuristic estimate h(n) and ignores the actual cost traveled so far. It may choose a node that looks close to the goal but actually has a long path.</p>
    </div>
    <h3><span class="badge">A4</span> Question 4 Sample Answer</h3>
    <div class="step-box">
        <p>Greedy BFS uses a <strong>Priority Queue</strong> ordered by h(n). This is important because it always selects the node that appears closest to the goal first, which is the core idea of the greedy approach.</p>
    </div>
    <h3><span class="badge">A5</span> Question 5 Sample Answer</h3>
    <div class="step-box">
        <p>B has a smaller heuristic value (h=6) than A (h=8). Greedy BFS always chooses the node with the smallest h value, so it explores B before A.</p>
    </div>
    <h3><span class="badge">A6</span> Question 6 Sample Answer</h3>
    <div class="step-box">
        <p><strong>Advantage:</strong> Greedy BFS is faster because it only computes h(n) and does not need to track g(n).</p>
        <p><strong>Disadvantage:</strong> Greedy BFS is not optimal and not complete. It can get stuck in loops or find a suboptimal path.</p>
    </div>
    <h3><span class="badge">A7</span> Question 7 Sample Answer</h3>
    <div class="step-box">
        <p>Yes, Greedy BFS can get stuck in infinite loops if the graph has cycles. We can prevent this by maintaining a <strong>visited list</strong> to avoid revisiting nodes.</p>
    </div>
    <h3><span class="badge">A8</span> Question 8 Sample Answer</h3>
    <div class="step-box">
        <p>Choose Greedy BFS when speed is more important than optimality, the heuristic is very accurate, and the search space is large. For example, in real-time games where a quick solution is needed even if it's not perfect.</p>
