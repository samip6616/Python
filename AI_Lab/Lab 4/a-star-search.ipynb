<h2 class="section-title">Part 1: Background</h2>
<div class="card">
    <h3><span class="badge">1</span> What is A* Search?</h3>
    <div class="theory-box">
        <p><strong>Definition:</strong> A* (A-Star) is an informed search algorithm that finds the shortest path between a start node and a goal node. It uses both the <strong>actual cost from the start</strong> (g(n)) and an <strong>estimated cost to the goal</strong> (h(n)).</p>
        <p><strong>Evaluation Function:</strong> f(n) = g(n) + h(n)</p>
        <ul>
            <li><strong>g(n):</strong> Actual cost from start node to node n</li>
            <li><strong>h(n):</strong> Heuristic estimate of cost from node n to the goal</li>
            <li><strong>f(n):</strong> Estimated total cost of the path through node n</li>
        </ul>
    </div>
    <div class="theory-box">
        <p><strong>Simple Analogy:</strong></p>
        <p>Imagine you are planning a road trip from Kathmandu to Pokhara. You use Google Maps that shows both the distance you have already driven (g) and the estimated distance remaining (h). This gives you the total estimated trip cost (f). A* works exactly like this &ndash; it considers both how far you've come and how far you have to go.</p>
    </div>
    <div class="theory-box">
        <p><strong>Key Characteristics:</strong></p>
        <ul>
            <li>Uses a <strong>Priority Queue</strong> (ordered by f(n) value &ndash; smallest first)</li>
            <li><strong>Complete</strong> &ndash; will find a solution if one exists</li>
            <li><strong>Optimal</strong> &ndash; finds the shortest path if h(n) is <strong>admissible</strong> (never overestimates)</li>
            <li>More efficient than uninformed search algorithms</li>
        </ul>
    </div>
    <div class="tree"><strong>🌲 Graph with Edge Costs and Heuristics:</strong>
        <pre>        S (h=10)

<h2 class="section-title">Part 2: Algorithm</h2>
<div class="card">
    <h3><span class="badge">2</span> A* Search Algorithm</h3>
    <div class="step-box">
        <p><strong>Algorithm Steps:</strong></p>
        <ol>
            <li>Start with the initial node. Add it to a priority queue with f = g + h (where g=0, h=heuristic value).</li>
            <li>Create a dictionary to store the <strong>best g(n)</strong> cost found so far.</li>
            <li>While the priority queue is not empty:
                <ol type="a">
                    <li>Remove the node with the <strong>smallest f(n)</strong> value.</li>
                    <li>If this node is the <strong>goal</strong>, return the path.</li>
                    <li>If this node has already been visited, <strong>skip</strong> it.</li>
                    <li>Mark the current node as <strong>visited</strong>.</li>
                    <li>Generate all <strong>neighbors</strong> of the current node.</li>
                    <li>For each neighbor, calculate <strong>new_g = g(current) + edge_cost</strong>.</li>
                    <li>If neighbor is not visited or new_g is <strong>better</strong> than previously recorded g(neighbor):
                        <ul>
                            <li>Update g(neighbor) = new_g</li>
                            <li>Calculate f(neighbor) = new_g + h(neighbor)</li>
                            <li>Add neighbor to the queue</li>
                        </ul>
                    </li>
                </ol>
            </li>
            <li>If the queue becomes empty, <strong>no solution exists</strong>.</li>
        </ol>
    </div>
    <div class="step-box">
        <p><strong>Pseudocode:</strong></p>
        <div class="code-box">
            <pre>function A_STAR(graph, start, goal, heuristic):
    queue = [(heuristic(start), start, 0, [start])]
    g_cost = {start: 0}
    visited = []

    while queue is not empty:
        sort queue by f(n)
        f_val, current, g_val, path = queue.pop(0)

        if current in visited:
            continue

        if current == goal:
            return path

        visited.append(current)

        for neighbor, edge_cost in graph[current]:
            if neighbor in visited:
                continue

            new_g = g_val + edge_cost

            if neighbor not in g_cost or new_g &lt; g_cost[neighbor]:
                g_cost[neighbor] = new_g
                new_f = new_g + heuristic[neighbor]
                new_path = path + [neighbor]
                add (new_f, neighbor, new_g, new_path) to queue

    return None  // No path found</pre>
        </div>
    </div>
</div>

<h2 class="section-title">Part 3: Code</h2>
<div class="card">
    <h3><span class="badge">3</span> Complete Python Code</h3>
    <div class="code-box">
        <pre># Lab: A* Search Algorithm
# Program by: [Your Name]
# Roll no: [Your Roll No]
# Step 1: Define the Graph (connections with edge costs)
graph = {
    'S': [('A', 5), ('B', 4)],
    'A': [('C', 3), ('D', 2)],
    'B': [('D', 2), ('E', 3)],
    'C': [('F', 2)],
    'D': [('F', 2), ('G', 1)],
    'E': [('G', 2)],
    'F': [],
    'G': []
}
# Step 2: Define Heuristic Values (estimated distance to goal G)
heuristic = {
    'S': 10,
    'A': 8,
    'B': 6,
    'C': 5,
    'D': 4,
    'E': 3,
    'F': 2,
    'G': 0
}
# Step 3: A* Search Algorithm
def a_star(start, goal):
    # Queue stores (f_value, node, g_value, path)
    queue = [(heuristic[start], start, 0, [start])]
    g_cost = {start: 0}  # Best g(n) found so far
    visited = []          # List of visited nodes

    print("=" * 70)
    print("A* SEARCH ALGORITHM")
    print("=" * 70)
    print()
    print("Start Node:", start)
    print("Goal Node:", goal)
    print()
    print("Step | Current Node | g(n) | h(n) | f(n) | Path")
    print("-----|--------------|------|------|------|-----------------")

    step = 1

    while queue:
        # Sort by f value (smallest first)
        queue.sort()
        
        # Get node with smallest f
        f_val, current, g_val, path = queue.pop(0)

        # Skip if already visited
        if current in visited:
            continue

        print(f"{step:4} | {current:12} | {g_val:4} | {heuristic[current]:4} | {f_val:4} | {' -&gt; '.join(path)}")
        step = step + 1

        # If goal found, return path
        if current == goal:
            print()
            print("=" * 70)
            print("GOAL FOUND!")
            print("=" * 70)
            print("Path:", ' -&gt; '.join(path))
            print("Total Cost:", g_val)
            return path

        # Mark current as visited
        visited.append(current)

        # Explore neighbors
        for neighbor, edge_cost in graph[current]:
            if neighbor in visited:
                continue

            new_g = g_val + edge_cost

            # Check if better path found
            if neighbor not in g_cost or new_g &lt; g_cost[neighbor]:
                g_cost[neighbor] = new_g
                new_f = new_g + heuristic[neighbor]
                new_path = path + [neighbor]
                queue.append((new_f, neighbor, new_g, new_path))

    print("Goal not found!")
    return None
    # Step 4: Run the algorithm
a_star('S', 'G')
</pre>

<h2 class="section-title">Part 7: Sample Answers (For Reference)</h2>
<div class="card">
    <h3><span class="badge">A1</span> Question 1 Sample Answer</h3>
    <div class="step-box">
        <p>The evaluation function of A* is f(n) = g(n) + h(n).</p>
        <p><strong>g(n):</strong> Actual cost from the start node to node n.</p>
        <p><strong>h(n):</strong> Heuristic estimate of the cost from node n to the goal.</p>
    </div>
    <h3><span class="badge">A2</span> Question 2 Sample Answer</h3>
    <div class="step-box">
        <p>A* found the path: <strong>S &rarr; B &rarr; D &rarr; G</strong></p>
        <p>Total cost: 4 + 2 + 2 = <strong>8</strong></p>
    </div>
    <h3><span class="badge">A3</span> Question 3 Sample Answer</h3>
    <div class="step-box">
        <p>An <strong>admissible heuristic</strong> never overestimates the true cost to reach the goal: h(n) &le; h*(n).</p>
        <p>It is important because when h(n) is admissible, A* is <strong>guaranteed</strong> to find the optimal (shortest) path. Example: Straight-line distance is admissible.</p>
    </div>
    <h3><span class="badge">A4</span> Question 4 Sample Answer</h3>
    <div class="step-box">
        <p>A* uses a <strong>Priority Queue</strong> ordered by f(n) = g(n) + h(n). This is important because it always expands the node with the smallest estimated total cost first, which helps find the optimal path efficiently.</p>
    </div>
    <h3><span class="badge">A5</span> Question 5 Sample Answer</h3>
    <div class="step-box">
        <p>A* uses f(n) = g(n) + h(n) (actual cost + estimated cost). Greedy BFS uses only f(n) = h(n) (only estimated cost).</p>
        <p>A* is <strong>optimal and complete</strong>; Greedy BFS is <strong>not optimal and not complete</strong>.</p>
    </div>
    <h3><span class="badge">A6</span> Question 6 Sample Answer</h3>
    <div class="step-box">
        <p>If h(n) overestimates the true cost (non-admissible), A* may still find a path, but it is <strong>not guaranteed</strong> to be optimal. The algorithm might discard the optimal path because it overestimates the remaining cost.</p>
    </div>
    <h3><span class="badge">A7</span> Question 7 Sample Answer</h3>
    <div class="step-box">
        <p><strong>Time Complexity:</strong> O(b^d) &ndash; exponential in the depth of the solution.</p>
        <p><strong>Space Complexity:</strong> O(b^d) &ndash; stores all generated nodes in memory.</p>
    </div>
    <h3><span class="badge">A8</span> Question 8 Sample Answer</h3>
    <div class="step-box">
        <p>Choose A* when you need the <strong>guaranteed shortest path</strong> and memory is not a constraint. For example, in navigation systems, route planning, and puzzle solving where optimality is important.</p>
    </div>
